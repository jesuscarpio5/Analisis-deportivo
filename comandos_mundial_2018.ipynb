```python
from statsbombpy import sb
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

EQUIPO = "Croatia"
partidos = sb.matches(competition_id=43, season_id=3)
partidos_eq = partidos[(partidos.home_team == EQUIPO) | (partidos.away_team == EQUIPO)]

eventos = pd.concat([sb.events(match_id=m) for m in partidos_eq.match_id], ignore_index=True)
eventos = eventos[eventos.team == EQUIPO]
print(len(partidos_eq), "partidos,", len(eventos), "eventos")
```

```python
columnas = ["match_id", "match_date", "home_team", "away_team", "home_score", "away_score", "competition_stage"]
print(partidos[columnas].sort_values("match_date").to_string(index=False))
```

```python
equipo = ""
partidos[(partidos.home_team == equipo) | (partidos.away_team == equipo)][
    ["match_id", "match_date", "home_team", "away_team", "home_score", "away_score", "competition_stage"]
].sort_values("match_date")
```
